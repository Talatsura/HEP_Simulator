```python
QC = qk.QuantumCircuit(3)

QC.ccx(0,1,2)

QC.cx(0,1)

QC.swap(1,2)

QC.cx(2,0)

QC.draw('mpl')
```

```python
psi = qk.quantum_info.Statevector.from_int(0b011,2**3)
print('Initial state:')
display(psi.draw('latex'))

psi = psi.evolve(QC)

print('Final state:')
display(psi.draw('latex'))
```
